```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root / "src"))

from material.twelve_tone import (
    generate_row,
    TwelveToneMaterial,
)

from processes.birth_death import (
    BirthDeathConfig,
    MutationConfig,
    BirthDeathToneRowProcess,
)

from rendering.midi import (
    pitch_events_to_midi,
    play_midi_in_notebook,
)

from analysis.stats import (
    pitch_events_to_dataframe,
    population_events_to_dataframe,
)


# -------------------------
# Parameters
# -------------------------

song_name = "birth_death_12tone_streamlit_prep"
midi_path = Path("../generated/midi") / f"{song_name}.mid"

row_seed = 420
process_seed = 69

bpm = 120
ticks_per_beat = 480
base_octave = 4
octave_span = 4

animation_step_seconds = 0.25
playback_speed = 1.0
animation_interval_ms = int(animation_step_seconds * 1000 / playback_speed)


# -------------------------
# 12-tone material
# -------------------------

raw_row = generate_row(modulus=12, seed=row_seed)
material = TwelveToneMaterial(raw_row=raw_row, normalise=True)

input_rows = [
    material.get_form("P", 0).row,
    material.get_form("I", 0).row,
    material.get_form("R", 0).row,
    material.get_form("RI", 0).row,
]


# -------------------------
# Process config
# -------------------------

bd_config = BirthDeathConfig(
    song_length=60.0,
    initial_population=2,
    min_population=2,
    max_population=12,
    birth_rate=0.12,
    death_rate=0.08,
    note_rate_mean=2.0,
    note_rate_sd=0.35,
    duration_mean=0.25,
    duration_sd=0.25,
    duration_mode="normal",
    duration_min=0.05,
    duration_max=1.50,
    velocity_mean=72,
    velocity_sd=0.12,
    velocity_mode="normal",
    seed=process_seed,
)

mutation_config = MutationConfig(
    transpose_probability=0.45,
    transpose_intervals=(-6, -5, -3, -2, 1, 2, 3, 5, 6),
    invert_probability=0.20,
    retrograde_probability=0.25,
    rotate_probability=0.25,
    rotate_steps=(1, 2, 3, 4, 5, 6),
)


# -------------------------
# Generate events and MIDI
# -------------------------

process = BirthDeathToneRowProcess(
    rows=input_rows,
    config=bd_config,
    mutation_config=mutation_config,
)

pitch_events, population_events = process.run()

saved_path = pitch_events_to_midi(
    pitch_events=pitch_events,
    output_path=midi_path,
    bpm=bpm,
    ticks_per_beat=ticks_per_beat,
    base_octave=base_octave,
    octave_span=octave_span,
)


# -------------------------
# DataFrames
# -------------------------

pitch_df = pitch_events_to_dataframe(pitch_events)
pop_df = population_events_to_dataframe(population_events)

frame_times = np.arange(
    0,
    bd_config.song_length + animation_step_seconds,
    animation_step_seconds,
)

population_at_frame = []

for t in frame_times:
    past_events = pop_df[pop_df["time"] <= t]

    if past_events.empty:
        population_at_frame.append(0)
    else:
        population_at_frame.append(int(past_events["population_size"].iloc[-1]))

anim_df = pd.DataFrame(
    {
        "time": frame_times,
        "population_size": population_at_frame,
    }
)


# -------------------------
# Summary
# -------------------------

print("Generated stochastic 12-tone birth-death MIDI")
print(f"MIDI path: {saved_path}")
print(f"Raw row: {material.raw_row}")
print(f"P0 row:  {material.p0_row}")
print()
print(f"Total notes:      {len(pitch_events)}")
print(f"Total voices:     {process.next_voice_id}")
print(f"Min population:   {anim_df['population_size'].min()}")
print(f"Max population:   {anim_df['population_size'].max()}")
print(f"Mean population:  {anim_df['population_size'].mean():.2f}")


# -------------------------
# Animation
# -------------------------

fig, ax = plt.subplots(figsize=(10, 4))

ax.set_xlim(0, bd_config.song_length)
ax.set_ylim(0, max(1, anim_df["population_size"].max() + 1))
ax.set_title("Birth-death active voices over time")
ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Active voices")
ax.grid(True, alpha=0.3)

line, = ax.step([], [], where="post")
point, = ax.plot([], [], marker="o")
time_text = ax.text(0.02, 0.92, "", transform=ax.transAxes)

def init():
    line.set_data([], [])
    point.set_data([], [])
    time_text.set_text("")
    return line, point, time_text

def update(frame_idx):
    x = anim_df["time"].iloc[: frame_idx + 1]
    y = anim_df["population_size"].iloc[: frame_idx + 1]

    line.set_data(x, y)
    point.set_data([x.iloc[-1]], [y.iloc[-1]])
    time_text.set_text(f"t = {x.iloc[-1]:.2f}s | voices = {y.iloc[-1]}")

    return line, point, time_text

animation = FuncAnimation(
    fig,
    update,
    frames=len(anim_df),
    init_func=init,
    interval=animation_interval_ms,
    blit=True,
)

plt.close(fig)
display(HTML(animation.to_jshtml()))


# -------------------------
# Audio preview
# -------------------------

play_midi_in_notebook(saved_path)
```
